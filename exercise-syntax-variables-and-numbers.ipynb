<small>Questions, like this one, marked a spicy pepper are a bit harder.</small>

Add parentheses to the following expression so that it evaluates to 0.

```python
nums = [8, 3, 2, 1, 2]
ops = ['-', '*', '-', '+']
target = 0

def parts(i, j):
    if i == j:
        return [str(nums[i])]
    out = []
    for k in range(i, j):
        for left in parts(i, k):
            for right in parts(k + 1, j):
                out.append(f"({left} {ops[k]} {right})")
    return out

for e in set(parts(0, len(nums) - 1)):
    if eval(e) == target:
        print(e)
```
